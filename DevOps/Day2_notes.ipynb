# **Version control Using Git**

# What is version control?
Version control is a system that records changes to a file or set of files over time so that you can recall specific versions later. It allows multiple people to work on a project simultaneously, tracks changes made by each contributor, and helps manage conflicts when merging changes.

# Problems befor version control

* Versioning was manually done by renaming files with version numbers or dates, which was error-prone and inefficient.

* Team Collaboration was difficult as multiple people working on the same file could lead to confusion and loss of work.

* No Easy access to history of changes made to a file, making it hard to track progress or revert to previous versions.

* Multiple versions of the same file could exist, leading to confusion and difficulty in identifying the most recent version. and occupying unnecessary storage space.

# Avantages of version control

* Version control allows for easy collaboration among team members, enabling multiple people to work on the same project simultaneously without overwriting each other's changes.

* Team Collaboration is improved as version control systems provide tools for managing conflicts and merging changes made by different contributors.

* Easy access to history of changes made to a file, allowing users to track progress, revert to previous versions, and understand the evolution of a project over time.

* Only modified files are stored, reducing storage space requirements and making it easier to manage large projects with many files.

# Types of version control systems





# Centeralized Version Control System (CVCS)
In a centralized version control system, there is a single central repository that contains all the files and their version history. Users check out files from the central repository, make changes, and then commit those changes back to the central repository. This model allows for easy collaboration and version tracking, but it also has some drawbacks, such as a single point of failure and limited offline access.

# Distributed Version Control System (DVCS)
In a distributed version control system, each user has a complete copy of the repository, including its entire history. Users can work independently, make changes, and commit them to their local repository. They can then synchronize their changes with other users by pushing and pulling changes to and from remote repositories. This model provides greater flexibility, offline access, and redundancy, as there is no single point of failure.

# What is Git?
Git is a distributed version control system that allows multiple developers to work on a project simultaneously. It tracks changes made to files, enables collaboration, and provides tools for managing conflicts and merging changes. Git is widely used in software development and is known for its speed, flexibility, and robust branching and merging capabilities.

* Git is a Dvcs, which means that each user has a complete copy of the repository, including its entire history. This allows for greater flexibility and offline access, as users can work independently and synchronize their changes with others when needed.

# Git Lifecycle
Git has three main states that your files can reside in: modified, staged, and committed.

* Modified: The file has been changed but not yet staged for commit.
* Staged: The file has been marked for commit and will be included in the next commit.
* Committed: The file has been saved to the local repository and is part of the project's history.
