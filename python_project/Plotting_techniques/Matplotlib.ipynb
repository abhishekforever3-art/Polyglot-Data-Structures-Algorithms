# Figure Class

## Def:
The Figure class is the overall container for all the elements in a Matplot.

it acts as a canvaas where multiple plots(Axes) can be placed.

Application:

Used when creating multiple subplots in a single figure.

helpfull for complex visuvalizations with different chatrs in one display.

```python
fig , ax = plt.subplots()
ax.set_title("Basic Figure and Axes")
ax.set_xlabel("Days")
ax.set_ylabel("Total patients")
plt.show()
```

# Line Plot 

Def:

a line plot displays points connected by lines. it is commonly used for showing trends over time .

application:

used for time series analysis 

in healthcare used to show patient recovery trends over weeks.





```python
months = ["Jan", "Feb", " mar", "Apr", "may", "jun"]
recovery_rate = [65, 70, 60, 78, 80, 85]

plt.plot(months, recovery_rate, marker='o', linestyle='-', color='b', label="Recovery Rate")

plt.xlabel("months")
plt.ylabel("recovery_percentage")
plt.title("Patient Recovery Trends")
plt.legend() # this is showing the top left box label in the graph.
plt.show()

```

# Bar plot 

def 

represents categorical data using rectangular bars, where the height is representing a numerical value.

applications

used to compare different categories 

in health care it can show number of pt in different hospital dept.

```python
medicines = ["Paracetamol", "Antibiotics", "Painkillers", "Insuline", " Antacids"]

demand = [500, 700, 800, 300, 250]

plt.bar(medicines, demand, color=['red', 'blue', 'orange', 'purple', 'green'], label= "Medicine Report")

plt.xlabel("Medicine type")
plt.ylabel('Units Sold')
plt.title("Medicine Demand Analysis")
plt.legend()
plt.show()
```

# Pie Chart 

def 

Shows proportions of different categories within a whole.

appli:

Used for percentage distribution 

in healthcare it can visuvalize % in distribution of disease and cases.

```python
diseases = ["Diabetes", "Hypertension", "Heart Disease","Asthma", "Others"]

percentage = [30, 40, 5, 10, 15]

colours = ['blue', 'green', 'red', 'purple', 'orange']

plt.pie(percentage, labels=diseases, autopct='%1.1f%%', colors=colours)

plt.title ("Disease Distribution in City")
#plt.legend()
plt.show()
```

# Scatter Plot

Shows individual data points on a 2d plain. it is used to identify relationship bt 2 variables.

used for correlation analysis.

in healthcare it can show corelation bt BMI and Blood Sugare

```python
time_Spent = [10, 20,15,25,30,35,40]

diagnosis_Accuracy = [60,65,70,75,80,85,90]

plt.scatter(time_Spent, diagnosis_Accuracy, color='red', label="Accuracy")

plt.xlabel("Time Spent by Doctor (minutes)")
plt.ylabel("Diagnosis Accuracy (%)")

plt.title ("Doctor- Patient Interaction Time Vs. Diagnosis Accuracy")
plt.legend()
plt.show()
```

# Word Cloud
is a visual representation of the text data where the size of each word is proportional to its frequency in data set.

used for text analysis.

in health it can analyse the pt feedback



```python
from wordcloud import WordCloud
Symptoms_text = "fever cough cough cold nausea headache cold headache nausea farigue pain weakness dizziness allergy infection sore throat"
wordcloud = WordCloud(width=800, height= 400, background_color= "blue").generate(Symptoms_text)

plt.figure(figsize=(10, 5))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')
plt.title("common Symptoms")
plt.show()
```
