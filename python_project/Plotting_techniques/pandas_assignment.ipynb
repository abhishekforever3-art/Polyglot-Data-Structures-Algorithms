# Assignment 2 - Working with Pandas

**Objective:** Learn to create and manipulate dataframes for data analysis.

## Step 1: Import Pandas
Import the pandas library for structured data manipulation.

```python
import pandas as pd
```

## Part 1: DataFrame Creation & Basic Analysis
Create the DataFrame using the provided data dictionary, print rows, calculate statistics, and find the average HR salary.

```python
# Create the DataFrame
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
    'Age': [24, 27, 22, 32, 29],
    'Department': ['HR', 'Finance', 'IT', 'Marketing', 'HR'],
    'Salary': [45000, 54000, 50000, 62000, 47000]
}

df = pd.DataFrame(data)

# a. Print the first five rows of the DataFrame
print("--- First Five Rows ---")
print(df.head())
print()

# b. Get the summary statistics of the 'Age' and 'Salary' columns
print("--- Summary Statistics (Age & Salary) ---")
print(df[['Age', 'Salary']].describe())
print()

# c. Calculate the average salary of employees in the 'HR' department
hr_avg_salary = df[df['Department'] == 'HR']['Salary'].mean()
print(f"Average Salary in HR Department: ${hr_avg_salary:,.2f}")
```

## Part 2: Add New Column
Add a 'Bonus' column equal to 10% of each employee's salary.

```python
# Add a new column, 'Bonus', which is 10% of the salary
df['Bonus'] = df['Salary'] * 0.10
print("DataFrame with Bonus Column:")
print(df)
```

## Part 3: Filtering Data
Filter the DataFrame to show employees aged between 25 and 30.

```python
# Filter the DataFrame to show employees aged between 25 and 30
filtered_df = df[(df['Age'] >= 25) & (df['Age'] <= 30)]
print("Employees Aged Between 25 and 30:")
print(filtered_df)
```

## Part 4: Grouping Data
Group the dataframe by Department to find average salary per department.

```python
# Group the data by 'Department' and calculate the average salary for each department
dept_avg_salary = df.groupby('Department')['Salary'].mean()
print("Average Salary by Department:")
print(dept_avg_salary)
```

## Part 5: Sorting and Exporting
Sort by salary ascending and export to a new CSV file.

```python
# Sort the DataFrame by 'Salary' in ascending order and save the result to a new CSV file
sorted_df = df.sort_values(by='Salary', ascending=True)
sorted_df.to_csv('sorted_employees.csv', index=False)
print("Sorted DataFrame (Saved to sorted_employees.csv):")
print(sorted_df)
```
