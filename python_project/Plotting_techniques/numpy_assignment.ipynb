## Step 1: Import NumPy
First, we import the NumPy library to use its array manipulation and mathematical functions.

```python
import numpy as np
```

## Part 1: 1D Array Operations
Create a 1D NumPy array with integers from 1 to 20. Calculate statistical metrics and find indices of elements greater than 10.

```python
# Create a 1D array with integers from 1 to 20
arr_1d = np.arange(1, 21)
print("1D Array:", arr_1d)

# a. Calculate the sum, mean, median, and standard deviation
arr_sum = np.sum(arr_1d)
arr_mean = np.mean(arr_1d)
arr_median = np.median(arr_1d)
arr_std = np.std(arr_1d)

print(f"Sum: {arr_sum}")
print(f"Mean: {arr_mean}")
print(f"Median: {arr_median}")
print(f"Standard Deviation: {arr_std:.2f}")

# b. Find the indices of elements greater than 10 in the array
indices_gt_10 = np.where(arr_1d > 10)
print("Indices where elements > 10:", indices_gt_10[0])
```

## Part 2: 2D Array Operations
Create a 2D NumPy array of shape 4x4 with numbers ranging from 1 to 16. Print it, find its transpose, and calculate row-wise and column-wise sums.

```python
# Create a 2D NumPy array of shape 4 X 4 with numbers ranging from 1 to 16
arr_2d = np.arange(1, 17).reshape(4, 4)

# a. Print the array
print("2D Array (4x4):\n", arr_2d)

# b. Find the transpose of the array
arr_transpose = arr_2d.T
print("\nTranspose of the Array:\n", arr_transpose)

# c. Calculate the row-wise and column-wise sums of the array
row_sums = np.sum(arr_2d, axis=1)
col_sums = np.sum(arr_2d, axis=0)
print("\nRow-wise Sums:", row_sums)
print("Column-wise Sums:", col_sums)
```

## Part 3: Random Arrays & Matrix Operations
Create two 3x3 arrays filled with random integers between 1 and 20. Perform element-wise operations and compute their dot product.

```python
# Create two 3 X 3 arrays filled with random integers between 1 and 20
rand_arr1 = np.random.randint(1, 21, size=(3, 3))
rand_arr2 = np.random.randint(1, 21, size=(3, 3))

print("Random Array 1:\n", rand_arr1)
print("\nRandom Array 2:\n", rand_arr2)

# a. Perform element-wise addition, subtraction, and multiplication
print("\nElement-wise Addition:\n", rand_arr1 + rand_arr2)
print("\nElement-wise Subtraction:\n", rand_arr1 - rand_arr2)
print("\nElement-wise Multiplication:\n", rand_arr1 * rand_arr2)

# b. Compute the dot product of the two arrays
dot_product = np.dot(rand_arr1, rand_arr2)
print("\nDot Product:\n", dot_product)
```

## Part 4: Reshaping and Slicing
Reshape a 1D array of size 12 into a 3x4 2D array and slice the first two rows and last two columns.

```python
# Reshape a 1D array of size 12 into a 3 X 4 2D array
arr_12 = np.arange(12)
arr_3x4 = arr_12.reshape(3, 4)
print("Reshaped 3x4 Array:\n", arr_3x4)

# Slice the first two rows and last two columns
sliced_arr = arr_3x4[:2, -2:]
print("\nSliced Array (First 2 rows, Last 2 columns):\n", sliced_arr)
```
