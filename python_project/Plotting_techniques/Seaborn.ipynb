# **Relational Plots**
* **Definition:**
* Relational plots show relationships between two or more numerical variables.


* **Usage:**
* Used to visualize relationships in datasets (e.g., correlation between age and medical expenses).
* Helps understand trends, patterns, and clusters in numerical data.


* **Characteristics:**
* Typically uses scatter plots (`sns.scatterplot`) and line plots (`sns.lineplot`).
* Supports `hue`, `size`, and `style` to add more insights.


* **Applications:**
* **Telecommunication:** Relationship between call duration and customer satisfaction.
* **Finance:** Relationship between income and expenditure.
* **Retail:** Impact of discount percentage on sales.



* **Telecommunication Case Study: Customer Call Analytics**
* **Problem Definition**
* The objective is to analyze **customer calling behavior** to understand trends in call durations, network complaints, and customer engagement patterns.


* **Dataset**
* **Call_Duration:** Duration of each call (in minutes)
* **Network_Provider:** Telecom provider (e.g., Provider A, Provider B)
* **Satisfaction_Score:** Customer's satisfaction score (1-10)


* **Relational Plot: Call Duration vs. Satisfaction Score**

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("telecom_customer_data.csv")
df.head()
```

```python
df.shape
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x="Call_Duration", y="Satisfaction_Score", hue="Network_Provider")


plt.legend(title="Network_Provider", bbox_to_anchor=(1,1),loc="upper left")

plt.title("Call Duration vs Satisfaction Score")
plt.show()
```


# **Categorical Plots**
* **Definition:**
* Categorical plots display relationships involving **categorical variables** (e.g., gender, department, product category).


* **Usage:**
* Used to analyze **grouped data** (e.g., average number of calls per customer segment).
* Helps in comparing different categories.


* **Characteristics:**
* Includes **bar plots** (`sns.barplot`), **box plots** (`sns.boxplot`), **violin plots** (`sns.violinplot`), etc.
* Allows **comparison across multiple groups**.


* **Applications:**
* **Telecommunication:** Distribution of customer complaints by network provider.
* **Healthcare:** Distribution of patient recovery time by disease.
* **Education:** Comparison of students' performance by subject.


* **Categorical Plot:** Count of Customers by Network Provider

```python
plt.figure(figsize=(6, 4))
ax = sns.countplot(
	data=df,
	x="Network_Provider",
	hue="Network_Provider",
	palette="Set1"
)

ax.set_title("Customer Distribution by Network Provider")
ax.set_xlabel("Network Provider")
ax.set_ylabel("Count of Customers")

# Remove the legend generated by hue, since x and hue are the same category.
if ax.legend_ is not None:
	ax.legend_.remove()

plt.show()
```


# **Distribution Plots**
* **Definition:**
* Distribution plots visualize the **spread** and shape of data.


* **Usage:**
* Helps understand **data distribution, skewness, and outliers**.
* Useful for **identifying trends and anomalies**.


* **Characteristics:**
* Includes **histograms** (`sns.histplot`), **kde plots** (`sns.kdeplot`), and **rug plots** (`sns.rugplot`).
* Displays **density estimates and probability distributions**.


* **Applications:**
* **Telecommunication:** Distribution of call durations per customer.
* **Finance:** Distribution of transaction amounts in a bank.
* **E-commerce:** Analyzing purchase frequency of products.

```python
plt.figure(figsize=(7,4))

sns.histplot(df['Call_Duration'], bins=20, kde=True, color='purple')

plt.title("Distribution of call Duration")

plt.xlabel("Call Duration (Minutes)")
plt.ylabel("Frequency")
plt.show()
```


# **Matrix Plots**
* **Definition:**
* Matrix plots visualize **relationships between multiple variables** using heatmaps and clustering methods.


* **Usage:**
* Helps find **correlations and patterns in datasets**.
* Used in **feature selection and anomaly detection**.


* **Characteristics:**
* Includes **heatmaps** (`sns.heatmap`) and **clustermaps** (`sns.clustermap`).
* Displays **intensity and similarity between variables**.


* **Applications:**
* **Telecommunication:** Heatmap of network congestion across different locations.
* **Finance:** Correlation between various stock indices.
* **Weather Analytics:** Heatmap of temperature variations over time.

```python
plt.figure(figsize=(8, 6))

numerical_df = df.select_dtypes(include=['number'])
sns.heatmap(numerical_df.corr(), annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)

plt.title("Correlation Matrix of Features")
plt.show()
```

# **Box Plot**

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='Network_Provider', y='Satisfaction_Score')
plt.xlabel("Network Provider")
plt.ylabel("Satisfaction Score")
plt.title("Satisfaction Score Across Different Network Providers")
plt.show()
```


**Inference:**

* Higher median satisfaction for a provider may indicate better service quality.
* Wide IQR (Interquartile Range) means user experiences vary greatly.
* Outliers suggest extreme positive or negative experiences.

# **Violin Plot: Call Duration by Network Provider**

* Swarmplots are useful for displaying individual data points, especially when dealing with categorical variables.

```python
plt.figure(figsize=(8, 5))
sns.violinplot(data=df, x='Network_Provider', y='Call_Duration')
plt.xlabel("Network Provider")
plt.ylabel("Call Duration (minutes)")
plt.title("Call Duration Distribution Across Network Providers")
plt.show()
```

**Inference:**

* Providers with wide distributions have highly varying call durations.
* Narrower shapes indicate more consistent call durations.
* Long tails suggest some customers have significantly longer or shorter calls.

**Conclusion:**

With these Seaborn visualizations, we gain deeper insights into our dataset, helping us understand relationships and distributions effectively.
