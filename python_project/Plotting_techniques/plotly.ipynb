## Plotly: Interactive Visualizations for Tourism & Travel Analytics

### Line Plot

#### Concept

* A **line plot** is used to visualize trends over time, making it ideal for **time-series data** like monthly tourist arrivals.

```python
import plotly.express as px
import pandas as pd

df = pd.read_csv("tourism_data.csv")
fig = px.line(df, x='Month', y='Tourist_Arrivals', title="Monthly Tourist Arrivals")
fig.show()
```


# **Inference:**

* The graph helps **identify seasonal trends**.
* Peaks may indicate **holiday seasons**, while dips suggest **off-seasons**.
* Between May to September is peak season for tourist.
* From October to December it is declining and lowest is in January.
* From January to July it is upward trend and August to December is downward trend.

* **Concept**
* A **scatter plot** helps visualize relationships between **two numerical variables**.


* **Hotel Bookings vs. Travel Cost**

```python
ab = px.scatter(df, x='Average_Travel_Cost', y="Hotel_Bookings", title='Hotel Booking vs.Travel Cost')
ab.show()
```

# **Histogram**

**Concept:**

# Violin Plot

### Concept

* A **violin plot** combines a **box plot** and **KDE** (Kernel Density Estimation) to show **data distribution with density estimation**.

### Seasonal Variation in Travel Cost

---

This provides a helpful guide on how to build and interpret violin plots in Python to visualize data distributions effectively.

```python
ab = px.violin(df, x='Season', y='Average_Travel_Cost', box= True, title='Seasonal Variation in Travel Cost')
ab.show()
```


## Inference

* Wider areas indicate **most frequent travel costs**.
* Outliers help **identify premium or budget travel trends**.
* **Highest Travel Cost**: Summer has the highest average travel cost at approximately 1000 units, making it the most expensive season for travel.
* **Lowest Travel Cost**: Winter has the lowest average travel cost at around 500 units, making it the most affordable season for travel.
* **Trend**: Travel costs rise steadily from Winter to Summer, peak in Summer, and then decline in Autumn, suggesting a cyclical pattern tied to seasonal demand (e.g., summer vacations increasing prices).


# **Gantt Chart**

* **Travel Itinerary Planning**

```python
ba = px.timeline(df, x_start= 'Start_Date', x_end='End_Date', y='Activity', title='Tourism Itinary')
ba.show()
```


**Inference:**

* **Diverse Options:** The list covers a wide range of interests (adventure, culture, relaxation), indicating a flexible or multi-purpose travel plan spanning 2025–2026.
* **Possible Gaps:** Months like Feb, Apr, Jun, etc., are missing, implying either no travel planned or focus on specific periods.

# Heatmap

### Concept

* A **heatmap** is useful for **finding patterns and correlations**.

```python
import plotly.figure_factory as ff
import numpy as np
```

```python
import plotly.graph_objects as go

# Compute correlation matrix (numerical columns only)
correlation_matrix = df.select_dtypes(include=['number']).corr()

# Create annotated heatmap using go.Heatmap
fig = go.Figure(data=go.Heatmap(
    z=correlation_matrix.values,
    x=correlation_matrix.columns.tolist(),
    y=correlation_matrix.index.tolist(),
    colorscale='Viridis',
    showscale=True,
    hoverongaps=False,
    text=[[f"{val:.2f}" for val in row] for row in correlation_matrix.values],
    texttemplate="%{text}",
))

# Update layout for spacing and readable text
fig.update_layout(
    title="Correlation Heatmap",
    width=800,
    height=800,
    margin=dict(t=50, l=100),
    font=dict(size=10)
)

fig.show()
```


## Inference

* **Inverse Relationship:** Higher `Hotel_Rating` (1.0) correlates with lower `Tourist_Arrivals` (0.098), suggesting premium-rated hotels may attract fewer, possibly high-end travelers.
* **Price vs. Demand:** The highest `Hotel_Price` (1.0) has mid-range `Hotel_Bookings` (0.461), indicating price sensitivity—luxury stays may not always maximize occupancy.
* **Travel Cost Impact:** Peak `Average_Travel_Cost` (1.0) aligns with high `Hotel_Bookings` (0.911), implying popular seasons (e.g., summer) drive both costs and demand.
* **Key Insight:** Balance between affordability (price/bookings) and exclusivity (rating/arrivals) is critical in tourism trends.

# 3D Plot

### Concept

* A **3D scatter plot** provides a **multi-dimensional view of relationships**.

### Travel Cost vs. Hotel Ratings & Bookings




```python
fig = px.scatter_3d(df, x='Average_Travel_Cost', y='Hotel_Rating', z='Hotel_Bookings',
                    title='3D Analysis of Travel Cost, Ratings & Bookings', color='Hotel_Rating')
fig.show()
```


## Inference

* Helps compare **luxury vs. budget travel trends**.
* Identifies clusters of **popular hotel price ranges**.
* **Positive Correlation Between Hotel Rating and Bookings:** Higher-rated hotels (e.g., 4.8–4.9) tend to have more bookings, indicating that customers prioritize quality and ratings when choosing accommodations.
* **Travel Cost Impact on Bookings:** Mid-range travel costs may correlate with higher bookings, as customers likely balance affordability with quality. Extremely high or low travel costs could show fewer bookings, suggesting a preference for value-for-money options.
* **Rating Consistency Matters:** Hotels with ratings around 4.4–4.5 appear frequently, suggesting this range is common or optimal for attracting bookings while maintaining competitive pricing. Lower ratings (e.g., 4.0–4.2) may indicate fewer bookings or less customer preference.
