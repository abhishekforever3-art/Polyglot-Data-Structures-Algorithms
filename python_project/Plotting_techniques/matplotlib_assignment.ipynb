# Assignment 3 - Working with Matplotlib

**Objective:** Practice data visualization techniques for better data representation.

## Step 1: Import Matplotlib & NumPy
Import pyplot and numpy to plot charts and generate sample data distributions.

```python
import matplotlib.pyplot as plt
import numpy as np
```

## Part 1: Line Plot
Plot a simple line chart with customized title, axis labels, and grid lines.

```python
x = [1, 2, 3, 4, 5]
y = [10, 15, 25, 30, 50]

plt.figure(figsize=(8, 5))
plt.plot(x, y, marker='o', color='b', linestyle='-', linewidth=2)
plt.title('Simple Line Plot Example', fontsize=14)
plt.xlabel('X Values', fontsize=12)
plt.ylabel('Y Values', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

## Part 2: Bar Graph
Represent student marks using custom-colored bar charts.

```python
students = ['John', 'Jane', 'Alice', 'Bob']
marks = [75, 85, 60, 90]

plt.figure(figsize=(8, 5))
plt.bar(students, marks, color=['skyblue', 'lightgreen', 'salmon', 'orange'])
plt.title('Student Marks in Subject', fontsize=14)
plt.xlabel('Students', fontsize=12)
plt.ylabel('Marks', fontsize=12)
plt.ylim(0, 100)
plt.show()
```

## Part 3: Pie Chart
Display company revenue distribution across regions with an exploded segment highlighting the highest earner.

```python
regions = ['North America', 'Europe', 'Asia', 'Others']
revenue = [45, 25, 20, 10]
explode = (0.1, 0, 0, 0)  # Highlight North America

plt.figure(figsize=(7, 7))
plt.pie(revenue, labels=regions, explode=explode, autopct='%1.1f%%', 
        shadow=True, startangle=140, colors=['gold', 'lightskyblue', 'lightcoral', 'lightgrey'])
plt.title('Percentage Distribution of Company Revenue by Region', fontsize=14)
plt.show()
```

## Part 4: Histogram
Generate a histogram showing frequency distribution of 1000 random integers between 1 and 100.

```python
random_data = np.random.randint(1, 101, size=1000)

plt.figure(figsize=(8, 5))
plt.hist(random_data, bins=10, color='purple', edgecolor='black', alpha=0.7)
plt.title('Frequency Distribution of Random Integers (1 to 100)', fontsize=14)
plt.xlabel('Value Range', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
