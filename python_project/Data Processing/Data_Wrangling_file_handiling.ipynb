### Data Wrangling & File Handling in Various Formats

* In real-world data analysis, information comes in different file formats like text (`.txt`), TSV (`.tsv`), and JSON (`.json`).
* Python provides built-in and library-based tools to read, write, and process these formats efficiently.
* This topic helps us learn how to load data from different sources and prepare it for analysis and modeling.

```python
with open('example_test.txt','w') as file:
    file.write('This is a sample text.\n')
print("file Created")
```


## Key Concepts:

* `open()`: Opens a file in write (`'w'`) or read (`'r'`) or other file mode.
* `write()`: Writes data to the file.
* `with statement`: Ensures proper file closure after operations.

```python
with open('example_test1.txt','w') as file:
    file.write('This is a sample text.\n')
    file.write('Second line of text.\n')
    
print("file Created")
```

```python
with open('example_test.txt','w') as file:
    for i in range(1,6):
        file.write(f"this is line{i} text.\n")
print("file created sucsessfully")
```

```python
with open ('example_test.txt', 'r')as file:
    data = file.read()
    print(data)
```


## Key Concepts:

* `read()`: Reads the entire file content.
* `readlines()`: Returns a list of lines (useful for multi-line files).

# Writting multiple lines

```python
lines = [
    "This is the first line.",
    "This is the second line.",
    "This is the third line.",
    "This is the fourth line.",
    "This is the fifth line."
]

with open('example_multi.txt', 'w') as file:
    for li in lines:
        file.write(li + "\n")
        
```

```python
with open('example_multi.txt','r') as file:
    lines = file.read()
for i in lines:
    print(i.strip()) # this will remove blank spaces
```


## Key Concepts:

* Absolute/relative paths for file access.
* `strip()`: Removes leading/trailing whitespace (including `\n`)


## Compatibility with Tools:

* TSV works seamlessly with spreadsheet software (Excel, Google Sheets), databases, and data analysis tools (Pandas, R).

## Avoiding Parsing Issues:

* TXT files with spaces/commas can cause ambiguity (e.g., "New York" vs. two columns).
* TSV's tabs are unambiguous delimiters.

## Efficiency:

* TSV is more compact than CSV for certain data (e.g., when text contains commas).

# Writting Df to a TSV file

```python
import pandas as pd

data = {
    'Emp_ID': [101, 102, 103, 104],
    'Name': ['Rahul Sharma', 'Priya Patel', 'Amit Singh', 'Neeta Rao'],
    'Department': ['HR', 'Finance', 'IT', 'Marketing'],
    'Salary (₹)': [50000, 65000, 80000, 55000],
    'Joining_Date': ['2020-01-15', '2019-05-22', '2021-11-10', '2022-03-05'],
    'City': ['Mumbai', 'Delhi', 'Bangalore', 'Hyderabad']
}

data
```

```python
df = pd.DataFrame(data)
df
```

```python
df.to_csv('table_example.tsv', sep='\t', index=False)
```


## Key Concepts:

* `sep='\t'`: Specifies tab as the delimiter.
* `index=False`: Avoids writing row indices (by default).
* `index=True`: Use if your DataFrame has a custom index (e.g., dates, employee IDs) that you want to preserve.

```python
import json

data = {
    "employees": [
        {"name": "John", "age": 30, "department": "Sales"},
        {"name": "Meera", "age": 25, "department": "Marketing"},
        {"name": "Rahul", "age": 35, "department": "HR"}
    ]
}

# Writing the dictionary to a JSON file
with open('json_example.json', 'w') as file:
    json.dump(data, file, indent=4)
```

reading Json

```python
import json
with open('json_example.json','r') as file:
    data = json.load(file)
    
print('Employees Information:')
for i in data['employees']:
    print(f"Name:{i['name']}, Age: {i['age']}, Department:{i['department']},")
```
