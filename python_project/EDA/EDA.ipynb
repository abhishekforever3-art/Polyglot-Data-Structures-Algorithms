# Detect Outliers Using IQR (Alternate method)

```python
# there are outlires in Wind column we will use median

median_value = data['Wind'].median()
print(median_value)
```

```python
# Replace outliers with median

data.loc[data['Wind']< lower_bound, 'Wind']= median_value
data.loc[data['Wind']> upper_bound, 'Wind']= median_value
```

```python
# to confirm outliers imputed

plt.figure(figsize=(8,4))
sns.boxplot(y=data['Wind'])
plt.title('Boxplot of Wind - After Imputation')
plt.show()
```


## What is a Dummy Variable?

* A **dummy variable** is a numeric (0/1) representation of **categorical data** in machine learning and statistical models.

## Why is it Required?

* Most algorithms **cannot understand text** — they need numbers.
* Dummy variables **convert categories into binary format**, allowing models to process them effectively.
* For example, a column `Gender` with values `Male` and `Female` becomes:



| Gender_Male | Gender_Female |
| --- | --- |
| 1 | 0 |
| 0 | 1 |


# Normalization of the data

### Normalization: The Basics

* Normalization is a technique used in data preprocessing to scale the values of numerical features to a specific range.
* This ensures that no feature dominates others purely due to its scale, which is especially important for machine learning algorithms sensitive to feature magnitudes (e.g., gradient descent, k-NN).
* Two common normalization techniques are **StandardScaler** and **MinMaxScaler**.

```python
#Normalization of data
from numpy import set_printoptions
from sklearn.preprocessing import MinMaxScaler
```

```python
data.values
```

```python
array = data.values
array
```

```python
scaler = MinMaxScaler(feature_range=(0,1))
rescaledX = scaler.fit_transform(array)
```

```python
set_printoptions(precision=2)
print(rescaledX[0:5,:])
```


# Key Differences Between StandardScaler and MinMaxScaler

| Feature | StandardScaler | MinMaxScaler |
| --- | --- | --- |
| **Purpose** | Centers and scales data (mean = 0, std = 1). | Scales data to a fixed range (e.g., [0, 1]). |
| **Effect** | Removes mean and variance differences. | Compresses data into the specified range |
| **Output Range** | Not fixed; depends on data distribution. | Fixed, typically [0, 1] or custom range. |
| **Sensitivity** | Affected by outliers but less so than MinMaxScaler. | Highly sensitive to outliers. |
| **Use Case** | For algorithms sensitive to variance or normality. | For algorithms needing uniform range of values. |

```python
# Standardize data (0 mean, 1 stdev)
from sklearn.preprocessing import StandardScaler
```

```python
array = data.values
scaler = StandardScaler()
scaler.fit(array)
rescaledX = scaler.transform(array)

# summarize transformed data

set_printoptions(precision=2)
print(rescaledX[0:5,:])
```
