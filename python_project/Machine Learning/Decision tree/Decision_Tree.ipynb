## Decision Tree: C 5.0

* C5.0 is one of the best implementations of the Decision Tree algorithm.
* It is widely used for classification tasks (predicting categories).
* C5.0 is not used for regression—it's purely for classification.

### Tree Building Methodology:

* The first question is which feature to select first?
* There are various measures to identify the best decision tree splitting candidate.
* C5.0 uses **entropy**, a concept borrowed from information theory that quantifies the randomness, or disorder, within a set of class values.

### How do you identify good features?

* **High Entropy:** Data sets with high entropy are very diverse and provide little information about other items that may also belong in the set, as there is no apparent commonality.
* **Splitting Goal:** The decision tree hopes to find splits that reduce entropy, ultimately increasing homogeneity within the groups.

## Entropy

* **Measurement Unit:** Typically, entropy is measured in **bits**.
* **Binary Classes:** If there are only two possible classes, entropy values can range from $0$ to $1$.
* **Multiple Classes:** For $n$ classes, entropy ranges from $0$ to $\log_2(n)$.
* **Interpretation of Values:**
* A **minimum value** indicates that the sample is completely homogeneous (pure).
* A **maximum value** indicates that the data are as diverse as possible, with no group having even a small plurality.



### Mathematical Formula

Entropy is computed using the formula:

$$\text{Entropy}(S) = \sum_{i=1}^{c} -p_i \log_2(p_i)$$

*(where $p_i$ represents the proportion of classes)*

## Information Gain

* **Definition:** To use entropy to determine the optimal feature to split upon, the algorithm calculates the change in homogeneity that would result from a split on each possible feature, a measure known as **information gain**.
* **Calculation:** The information gain for a feature $F$ is calculated as the difference between the entropy in the segment before the split ($S_1$) and the partitions resulting from the split ($S_2$):

$$\text{InfoGain}(F) = \text{Entropy}(S_1) - \text{Entropy}(S_2)$$

* **Root Node Selection:** The feature that yields the most information gain will be considered as the root node.
* **Weighted Entropy of Partitions:** After splitting the feature, the function to calculate $\text{Entropy}(S_2)$ needs to consider the total entropy across all of the partitions, weighted by their proportions ($w_i$):

$$\text{Entropy}(S) = \sum_{i=1}^{n} w_i \text{Entropy}(P_i)$$

To calculate the value of the expression $-\frac{3}{5}\log_2\left(\frac{3}{5}\right) - \frac{2}{5}\log_2\left(\frac{2}{5}\right)$, follow these steps:

1. **Compute $\log_2\left(\frac{3}{5}\right)$**:

$$\log_2\left(\frac{3}{5}\right) = \log_2(3) - \log_2(5)$$


2. **Compute $\log_2\left(\frac{2}{5}\right)$**:

$$\log_2\left(\frac{2}{5}\right) = \log_2(2) - \log_2(5) = 1 - \log_2(5)$$


3. **Substitute these values into the expression**:

$$-\frac{3}{5}(\log_2(3) - \log_2(5)) - \frac{2}{5}(1 - \log_2(5))$$


4. **Simplify the expression**:

$$= -\frac{3}{5}\log_2(3) + \frac{3}{5}\log_2(5) - \frac{2}{5} + \frac{2}{5}\log_2(5)$$


$$= -\frac{3}{5}\log_2(3) + \left(\frac{3}{5} + \frac{2}{5}\right)\log_2(5) - \frac{2}{5}$$


$$= -\frac{3}{5}\log_2(3) + \log_2(5) - \frac{2}{5}$$


5. **Evaluate each term**:

$$\log_2(3) \approx 1.58496$$


$$\log_2(5) \approx 2.32193$$



Substitute these values back into the expression:

To calculate the value of $-\frac{3}{5}\log_2\left(\frac{3}{5}\right) - \frac{2}{5}\log_2\left(\frac{2}{5}\right)$, you can follow these steps:

1. **Calculate $\log_2\left(\frac{3}{5}\right)$**:

$$\log_2\left(\frac{3}{5}\right) = \log_2(3) - \log_2(5)$$


2. **Calculate $\log_2\left(\frac{2}{5}\right)$**:

$$\log_2\left(\frac{2}{5}\right) = \log_2(2) - \log_2(5) = 1 - \log_2(5)$$


3. **Substitute these values into the expression**:

$$-\frac{3}{5}(\log_2(3) - \log_2(5)) - \frac{2}{5}(1 - \log_2(5))$$


4. **Simplify the expression**:

$$= -\frac{3}{5}\log_2(3) + \frac{3}{5}\log_2(5) - \frac{2}{5} + \frac{2}{5}\log_2(5)$$


$$= -\frac{3}{5}\log_2(3) + \left(\frac{3}{5} + \frac{2}{5}\right)\log_2(5) - \frac{2}{5}$$


$$= -\frac{3}{5}\log_2(3) + \log_2(5) - \frac{2}{5}$$


5. **Evaluate each term**:

$$\log_2(3) \approx 1.58496$$


$$\log_2(5) \approx 2.32193$$



Substitute these values back into the expression:


$$= -\frac{3}{5} \cdot 1.58496 + \log_2(5) - \frac{2}{5}$$

$$= -\frac{3}{5} \cdot 1.58496 + 2.32193 - \frac{2}{5}$$

### 6. Calculate the Final Values

* **First term evaluation:**

$$-\frac{3}{5} \cdot 1.58496 \approx -0.950976$$


* **Third term evaluation:**

$$-\frac{2}{5} \approx -0.4$$



---

### Substitution and Final Result

Substitute these computed values back into the expression:

$$-0.950976 + 2.32193 - 0.4$$

$$= 0.970954$$

This completes the math, confirming our initial entropy value of **$0.97$** bits for the dataset before the split!

### Gini Impurity for Classification Problems

* **The Intuitive Analogy:** Imagine you're trying to help a friend decide whether to buy a car or not, based on factors like price, mileage, and age of the car.
* **The Decision Process:** You ask simple **'yes' or 'no' questions**, one after another, until you arrive at a decision.
* **CART Framework:** That’s the basic idea behind **Decision Trees**—specifically one of the most popular types: **CART (Classification and Regression Trees)**, which uses **Gini Impurity** to determine optimal splits instead of entropy and information gain.

## What is CART?

* CART stands for **Classification and Regression Trees**.
* The beauty of CART is that it works for both:
1. **Classification problems** (predicting categories)
2. **Regression problems** (predicting continuous values)


* Think of CART as a flowchart-like structure where:
* Each **internal node** represents a decision on a feature.
* Each **leaf node** represents the final prediction.

## CART for Classification: Enter Gini Impurity

* **The Core Need:** When we use CART for classification (e.g., spam vs. not spam), we need a way to measure how **pure or impure** a node is.
* **The Solution:** That's where **Gini Impurity** comes in.

### What is Gini Impurity?

* **The Analogy:** Imagine you're blindfolded and picking a ball from a bag.
* **Pure Node:** If the bag has balls of **only one color**: you're 100% correct every time $\rightarrow$ **Gini Impurity = 0** (pure node).
* **Impure Node:** If the bag has **mixed colors**, you have a chance of being wrong $\rightarrow$ Gini Impurity increases (impure node).


## CART for Classification: Enter Gini Impurity

### The Gini Impurity Formula

$$\text{Gini} = 1 - \sum p_i^2$$

*(Where $p_i$ is the proportion of class $i$ in the node)*

---

### Step-by-Step Example Calculation

Suppose a node has:

* **$60\%$ Class A** $\rightarrow p_1 = 0.6$
* **$40\%$ Class B** $\rightarrow p_2 = 0.4$

$$\text{Gini} = 1 - (0.6^2 + 0.4^2)$$

$$\text{Gini} = 1 - (0.36 + 0.16)$$

$$\text{Gini} = 1 - 0.52$$

$$\text{Gini} = \mathbf{0.48}$$

---

### Interpretation

👉 **So, there's a 48% chance of incorrect classification** if we randomly assign labels to items in this node based on their class distribution.

## CART for Classification: Enter Gini Impurity

### How CART Uses Gini Impurity

* **Split Selection:** CART looks for the split that **reduces Gini Impurity the most**.
* **Decision Process:** Each possible split acts like a potential question we could ask.
* **Optimization Goal:** The main objective is to find the question that makes the child nodes **purer** than the parent node.

> **Remember:** Lower Gini = Better split!

## CART for Regression: Enter Least Squares

* **The Limitation:** In regression (predicting numbers), we can't use Gini Impurity.
* **The Solution:** Instead, CART uses the idea of **Residual Sum of Squares (RSS)**.
* **Core Concept:** In each node, we compute the **mean value** of the target.
* **Residual Measurement:** We then measure the squared difference between the actual values and this mean (the residuals).
* **Optimization Goal:** CART looks for splits that **minimize these residuals**.

---

### In Simple Words:

> "In regression, the tree is trying to keep similar numbers together—so the average of the group is close to all the values inside it."

## Limitations of Decision Trees

### 1. Overfitting

* **Complexity and Noise:** Decision Trees can become **too complex**, capturing **noise** instead of the true pattern.
* **Generalization Issue:** A tree that's too deep may work perfectly on training data but perform poorly on new, unseen data.

---

### Analogy

> It’s like memorizing answers for an exam instead of actually understanding the concepts.

## Limitations of Decision Trees

### 2. Instability (High Variance)

* **Sensitivity to Data Changes:** Small changes in data can result in **completely different trees**.
* **Lack of Robustness:** This makes decision trees sensitive and less robust.

---

### 3. Bias Toward Features with More Levels

* **Preference for High Cardinality:** Decision Trees can favor features with **many unique values** (e.g., ID numbers), even if they're not meaningful.

## Pruning: The Solution to Overfitting

* **What is Pruning?** Pruning means cutting back the tree to remove parts that don't add much predictive value.

---

### Two Types of Pruning:

1. **Pre-Pruning (Early Stopping):**
* Stop the tree from growing beyond a certain **depth** or when the **number of samples** in a node is too small.


2. **Post-Pruning (Reduced Error or Cost Complexity):**
* First grow the full tree, then **remove branches** that don't significantly improve performance.
