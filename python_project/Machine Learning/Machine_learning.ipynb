### 1 What is Machine Learning?

* **Definition:**
* Machine Learning is a method of teaching computers to make decisions or predictions based on data, without explicitly programming every rule.


**Examples:**

* Healthcare: Predict whether a patient will be readmitted based on medical history.
* Education: Predict whether a student will pass based on attendance and assignment scores.
* HR: Predict if an employee will leave the company (attrition).

---

### 2 Types of Machine Learning

**Imagine this:**

* Credit card fraud alerts—your bank spots shady transactions by learning from past examples of fraud. Again, supervised learning.

* Netflix grouping users with similar tastes without any labels—that's unsupervised learning.
* A self-driving car improving lane changes through trial and error—that's reinforcement learning

## 1. Supervised Learning

**Definition:**

"Supervised learning uses *labeled data* (input-output pairs) to train models, like a teacher-student setup."


## Examples:

* **Banking:** Predicting loan defaults (e.g., features = salary, credit score; label = 'approved' or 'rejected').
* **Case Study:** LendingTree uses this to assess risk.
* **Healthcare:** Diagnosing diseases from X-rays (input = image; label = 'cancer' or 'normal').

> "Here, the model learns patterns from past data to predict future outcomes—just like how Netflix recommends shows based on your past watches!"

## 2. Unsupervised Learning

### Definition:

> "No labels here! The model finds hidden patterns in data, like a detective solving a mystery."

### Examples:

* **Retail:** Customer segmentation (e.g., grouping shoppers by purchase behavior).


* *Case Study*: Amazon clusters users for personalized ads.
* Genomics: Identifying gene patterns linked to diseases without pre-defined categories.


**Examples:**

* Gaming: AlphaGo mastered Go by playing millions of games.
* Robotics: Boston Dynamics' robots learn to walk without falling.

> "Every time the robot stays upright, it gets a 'reward'—slowly mastering the task!"


**Examples:**

* Email Filtering: Gmail classifies emails (e.g., 'spam' labels + unlabeled data).
(100,000 emails (only 1,000 are labeled as *spam* or *not spam*)).
* Medical Imaging: Labeled tumor images + unlabeled scans improve diagnostics.


### 3 Key Terminology in ML

| Term | Meaning |
| --- | --- |
| **Feature** | Input variable (e.g., Age, Education level) |
| **Target** | What you want to predict (e.g., Purchase made: Yes/No) |
| **Label** | Known target in supervised learning |
| **Training data** | Data used to build the model |
| **Testing data** | Data used to evaluate the model |
| **Model** | The learned formula or rules |


### Use-case Examples:

* **Finance**: Predict credit card default $\rightarrow$ **Features**: income, loan history $\rightarrow$ **Target**: defaulted?
* **Food Delivery**: Predict delivery time $\rightarrow$ **Features**: distance, weather $\rightarrow$ **Target**: time in mins

### 5 What Makes a Good Model?

* 🟩 **Accuracy**: Does the model make good predictions? *(highlighted)*
* ⚖️ **Bias vs Variance**: *(circled)*
* **High bias** $\rightarrow$ underfitting (too simple)
* **High variance** $\rightarrow$ overfitting (too complex)



**Storytelling Examples:**

* 🏥 Overfit model in cancer detection sees too many patterns $\rightarrow$ flags healthy people
* 🧪 Underfit model for chemical reaction predicts same result regardless of conditions

* Show underfit, good fit, overfit using simple curved and linear plots.
