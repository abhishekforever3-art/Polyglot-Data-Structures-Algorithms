## 🎓 Introduction to TF-IDF
What is TF-IDF?

In Natural Language Processing, not all words are equally important.
Words like 'the', 'is', or 'and' appear in almost every document — they
don't help us identify what a specific document is really about.
That's where TF-IDF comes in.

TF-IDF stands for Term Frequency – Inverse Document Frequency,
and it's a technique to find out how important a word is in a document
compared to a whole collection (called corpus).

Definition:
TF-IDF is a statistical measure used to evaluate how important a word is
to a document in a collection of documents.

🧰 Simple Analogy:
Think of your document collection like a library. If a word shows up in
only one book, it's likely specific and important to that book. If it shows
up in every book, it's probably just a common filler word.
TF-IDF helps us tell the difference.

Real-World Use Cases of NLP

💬 Key Message:

TF-IDF helps highlight *unique and meaningful words* while downplaying the
common ones.

📖 Why Do We Need TF-IDF?
One might wonder can't we just count how often each word appears in a
document and use that as a feature? That's called Term Frequency (TF).
But here's the problem: some words occur frequently across all documents,
like 'the', 'are', or 'data'. These don't tell us anything unique about a specific
document.

So, just counting words isn't enough. We need a smarter way — and that's
where TF-IDF helps!"

🎯 Real-Life Example:

| Document | Word | TF | IDF | TF-IDF |
| :--- | :--- | :--- | :--- | :--- |
| Doc 1 | "cancer" | 3 | High | High |
| Doc 2 | "the" | 10 | Low | Low |

✅ How TF-IDF Solves It:
* Boosts important terms (unique to fewer documents)
* Penalizes common terms (appear everywhere)
* Brings balance between frequency and rarity

TF-IDF helps focus on words that are both frequent in a document
and rare across documents — those are usually the most
meaningful.

 Alternatives to TF-IDF

| Method | Description | When to Use |
| :--- | :--- | :--- |
| Bag of Words (BoW) | Count word occurrences in a document (no weighting). | Simple tasks, small datasets |
| Word Embeddings | Represent words as vectors using models like Word2Vec, GloVe, FastText. | When meaning/semantic similarity matters |
| TF Only | Just counts how often a term appears. | Very basic models, or when speed matters |
| Transformer Embeddings | Context-aware vectors from models like BERT. | Advanced NLP (e.g., summarization, QA) |

💬 TF-IDF vs Alternatives

| Metric | BoW (Bag of words) | TF-IDF | Embeddings |
| :--- | :--- | :--- | :--- |
| Word Meaning | ❌ | ❌ | ✅ |
| Word Weighting | ❌ | ✅ | ✅ |
| Context Aware | ❌ | ❌ | ✅ |
| Easy to Implement | ✅ | ✅ | ❌ (needs training or pre-trained models) |

💬 Key Message:

TF-IDF is a sweet spot between simplicity and intelligence — it's fast,
interpretable, and powerful for many real-world tasks.

🎓 The Math Behind TF-IDF

* Term Frequency (TF) — how often a word appears in a document
* Inverse Document Frequency (IDF) — how rare the word is across documents

📊 Step-by-Step Breakdown

🟦 1. Term Frequency (TF)

How frequent is the word in the current document?

$$TF(t) = \frac{\text{Number of times term } t \text{ appears in the document}}{\text{Total number of terms in the document}}$$

Example:

Document = "Cancer is a serious disease. Cancer treatment is expensive."

Term = "cancer"

TF = 2 / 8 = 0.25

🟨 2. Inverse Document Frequency (IDF)

How rare is the word across all documents?

$$IDF(t) = \log\left(\frac{\text{Total number of documents}}{\text{Number of documents containing the term } t}\right)$$

Example:

* Total documents = 100
* "cancer" appears in 5 documents

$$IDF = \log(100/5) = \log(20) \approx 1.30$$

🟥 3. TF-IDF

$$TF-IDF(t) = TF(t) \times IDF(t)$$

So for the term "cancer":

$$TF-IDF = 0.25 \times 1.30 = 0.325$$

📌 Interpretation:

* Higher TF-IDF value $\rightarrow$ Word is **important and specific** to the document
* Lower TF-IDF value $\rightarrow$ Word is **too common or not important**

💬 Key Message:

TF-IDF is just a product of **how often** a term appears and **how rare** it is.

🎓 Python Code Example for TF-IDF

Use TfidfVectorizer from scikit-learn, which does all the math for us
behind the scenes.

**Refer python program to implement TF-IDF**

# Python Implementation of TF-IDF (Term Frequency – Inverse Document Frequency) 

### Step 1: Import necessary libraries

```python
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd
```

### Step 2: Define a small corpus (3 documents)

```python
docs = [
    "Cancer is a deadly disease",
    "Cancer treatment is expensive",
    "Regular exercise prevents disease"
]

```

### Step 3: Initialize TF-IDF Vectorizer

```python
vect = TfidfVectorizer()
```

### Step 4: Fit and transform the text

```python
X = vect.fit_transform(docs)
```

```python
X
```

### Step 5: Convert to DataFrame for better readability

```python
df = pd.DataFrame(X.toarray(), columns=vect.get_feature_names_out())
```

### Step 6: Display TF-IDF values

```python
print(df.round(2))
```

◆ Step 3: TF $\times$ IDF = TF-IDF

✅ **In Doc1:** "Cancer is a deadly disease"

| Term | TF | IDF | TF-IDF |
| --- | --- | --- | --- |
| cancer | 0.20 | 1.00 | $0.20 \times 1.00 = 0.20$ |
| deadly | 0.20 | 1.405 | $0.20 \times 1.405 = 0.28$ |

### 🔍 What You Can Observe 

Words like "cancer" appear in multiple documents

🔸 Even though "cancer" is an important word, its IDF is lower because it's not rare across the corpus.

🔸 This results in a moderate TF-IDF score.

Words like "deadly", "expensive", and "prevents" appear only once

🔸 These are unique to individual documents — making them highly specific.

🔸 Their IDF is high, so even with a low TF, the TF-IDF score is high.
