Introduction to NLP (Natural Language Processing)

**What is NLP?**

Natural Language Processing (NLP) is the field of AI that enables computers to understand and process human language.
It combines linguistics and machine learning to extract meaningful information from text.

**Why is NLP Important?**

We generate huge amounts of text daily, from social media posts and emails to news articles and chat messages.
Analyzing such large text data manually is impractical.

NLP automates this process for various real-world applications.

Key Steps in Text Preprocessing

1. Tokenization — Breaking text into words or sentences.
2. Stopword Removal — Removing common words that don't add meaning (e.g., "the", "is", "and").
3. Lemmatization & Stemming — Converting words to their base form ("running" $\rightarrow$ "run").
4. Part-of-Speech Tagging — Identifying grammatical roles (noun, verb, adjective).
5. Named Entity Recognition (NER) — Recognizing proper names, places, organizations.

What are Stop Words?

* Stop words are common words in a language that usually don't add much meaning.
* Examples: "is", "the", "and", "was", "have", "not".

Why remove them?

* They occur very frequently in all documents.
* They don't help in distinguishing between topics or classes in many NLP tasks.
* Removing them reduces noise and makes models faster.

But Be Careful:

Not always safe to remove stop words:

* In sentiment analysis, removing "not" can flip the meaning!

"This is not good." $\rightarrow$ becomes "good" if "not" is removed.

🪓 Stemming

**What is Stemming?**

* A text normalization technique that reduces inflected or derived words to their word stem.
* It uses rule-based chopping of suffixes.

🔤 Example:

* "running" $\rightarrow$ "run"
* "flies" $\rightarrow$ "fli"
* "responsibilities" $\rightarrow$ "respons"

🧪 Popular Stemmer in NLTK: Porter Stemmer

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()
print(stemmer.stem("running"))      # run
print(stemmer.stem("responsibilities")) # respons
print(stemmer.stem("flies"))      # fli

```

🔧 Characteristics

| Feature | Stemming |
| --- | --- |
| Speed | Very fast |
| Output | May not be a real word |
| Rule-based | Yes (no grammar knowledge) |
| Accuracy | Lower |

⚠️ Drawbacks

* "better" $\rightarrow$ "better" (no reduction)
* "universities" $\rightarrow$ "univers" (cut too much)
* "caring" $\rightarrow$ "care" ✅ BUT "careful" $\rightarrow$ "care" ❌

🧠 Analogy

Stemming is like breaking down a LEGO tower with a hammer — it works fast but isn't gentle. You may lose important pieces.


Lemmatization

**What is Lemmatization?**

* A smarter version of stemming.
* It reduces words to their dictionary root (lemma), considering the part of speech (POS).

**Examples:**

* "running" (verb) $\rightarrow$ "run"
* "better" (adjective) $\rightarrow$ "good"
* "caring" $\rightarrow$ "care"

✅ How to Do It (Using WordNet Lemmatizer in NLTK)

```python
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')
nltk.download('omw-1.4')

lemmatizer = WordNetLemmatizer()
print(lemmatizer.lemmatize("running", pos="v"))  # run
print(lemmatizer.lemmatize("better", pos="a"))   # good

```

---

🔧 Characteristics

| Feature | Lemmatization |
| --- | --- |
| Output | Always a valid dictionary word |


Requires POS? Yes, for best results

Uses Grammar: Yes (WordNet, POS rules)

Speed: Slower than stemming

Accuracy: High

---

**Analogy**

Lemmatization is like sending the word to a dictionary editor — they check grammar, part of speech, and return the correct base word.

🧪 Stemming vs Lemmatization

| Word | Stemmed | Lemmatized (POS: Verb) |
| --- | --- | --- |
| running | run | Run |
| better | better | good (if POS='a') |
| responsibilities | respons | Responsibility |
| studies | studi | Study |

---

🎯 When to Use What?

| Scenario | Use |
| --- | --- |
| Fast & dirty preprocessing | Stemming |

| Scenario | Use |
| --- | --- |
| Precision-focused applications (QA, Chatbots) | Lemmatization |
| Short text cleanup | Stop word removal |

````python
lemmatizer = WordNetLemmatizer()
print(lemmatizer.lemmatize("running", pos="v"))  # run
print(lemmatizer.lemmatize("better", pos="a"))   # good
print(lemmatizer.lemmatize("flies",pos='v'))   # fly

# pos="v" → Specifies the part of speech (verb here) for accurate lemmatization (default is noun, "n").
````

````python
# Full example: Stopword removal + Lemmatization
from nltk.tokenize import word_tokenize

from nltk.corpus import stopwords

# Download if not already
nltk.download('stopwords')

# Get English stop words
stop_words = stopwords.words('english')

text = "Cats are running around the house, chasing mice happily."

tokens = word_tokenize(text)
filtered = [word for word in tokens if word.lower() not in stop_words]

lemmatized = [lemmatizer.lemmatize(word.lower(), pos='v') for word in filtered]
print("Processed:", lemmatized)
````
