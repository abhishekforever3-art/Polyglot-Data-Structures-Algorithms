## University Clustering Analysis: Identifying Peer Groups Using K-Means Clustering

### Business Problem Understanding

In the higher education sector, universities frequently need to identify their **"peer institutions"** for effective benchmarking. Establishing these peer groups helps institutions with:

* **Comparing performance metrics** (e.g., graduation rates, student-faculty ratios).
* **Setting realistic strategic goals** (e.g., improving admissions selectivity or funding).
* **Understanding competitive positioning** in rankings and student recruitment.

---

### Objective

The primary goal is to cluster universities into distinct groups based on key institutional characteristics, categorized as follows:

* **Admissions Selectivity:** Metrics such as SAT scores, acceptance rates (`Accept`), and top 10% student standing (`Top10`).
* **Financial Resources:** Metrics such as instructional expenses (`Expenses`) and student-faculty ratios (`SFRatio`).
* **Student Success:** Metrics such as graduation rates (`GradRate`).

## Dataset Description

The dataset contains 25 universities with the following features:

* **Univ:** University name
* **SAT:** Median SAT score of admitted students
* **Top10:** Percentage of students in the top 10% of their high school class
* **Accept:** Acceptance rate (%)
* **SFRatio:** Student-to-faculty ratio
* **Expenses:** Instructional expenditure per student ($)
* **GradRate:** Graduation rate (%)
