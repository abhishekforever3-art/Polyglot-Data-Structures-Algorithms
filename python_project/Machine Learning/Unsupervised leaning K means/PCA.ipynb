## Dimensionality Reduction – PCA (Principal Component Analysis)

### Introduction

As we transition from clustering techniques, we often encounter high-dimensional datasets with many features. **Principal Component Analysis (PCA)** is a powerful unsupervised technique used to reduce the dimensionality of data while retaining as much variance and essential information as possible.


## Definition of PCA (Principal Component Analysis)

* PCA (Principal Component Analysis) is an **unsupervised linear transformation technique** used for dimensionality reduction.
* It projects high-dimensional data into a smaller set of new features (called **principal components**) that retain most of the important information (variance).
* **PCA is an Unsupervised Machine Learning technique** because it does not use class labels (y-values) during computation.
* It only looks at the structure of input features ($X$) to reduce dimensions.

## PCA Concept and Scree Plot

### Concept

* **Correlated Features:** Imagine data with many correlated features (e.g., height, weight, BMI). PCA finds new **uncorrelated axes** (principal components) that capture maximum variance.
* **First Principal Component:** Represents the direction of maximum variance in the dataset.
* **Second Principal Component:** Represents the next orthogonal direction capturing the maximum remaining variance.
* **Iteration:** This process continues until all principal components are extracted.

### Scree Plot

* **Definition:** A line plot of eigenvalues (representing variance explained) versus the number of components.
* **Purpose:** Helps decide how many principal components to keep in order to retain enough information while reducing dimensions.
* look for the "**elbow point**" ( where adding more components give diminishing returns)

## Why PCA is Used

* Reduce high-dimensional data into fewer dimensions for easier visualization.
* Remove multicollinearity (features highly correlated).
* Speed up machine learning training.
* Reduce overfitting by eliminating noise.

### Simple Example

* Imagine we collect **Height (cm)** and **Weight (kg)** of people.
* These two features are correlated (taller $\rightarrow$ heavier).
* PCA finds a new axis (a diagonal line) that explains most variance.
* Instead of storing two correlated values, we can keep one principal component (linear combination of height & weight) that captures most info.

## How PCA works conceptually

1. **Find directions of maximum variance** (principal components).
* The **1st principal component** is the line/axis that captures the greatest spread in the data.
* The **2nd component** is perpendicular (uncorrelated) and captures the next highest variance.


2. **Rank components by importance.**
* Each component has a "score" (called **explained variance**).
* Higher score = more important = captures more information.


3. **Decide how many components to keep.**
* Plot variance explained using a **Scree Plot**.
* Keep enough components to capture roughly $90\text{--}95\%$ of the total variance.


## Example: Height vs. Weight PCA

* **Diagonal Cloud:** Imagine plotting "height" versus "weight" of people; most points form a diagonal cloud $\rightarrow$ the main variation is along that diagonal.
* **Principal Component 1 ($PC1$):** PCA sets $PC1$ as this diagonal line, which explains roughly $95\%$ of the variance.
* **Principal Component 2 ($PC2$):** $PC2$ is the perpendicular line capturing the tiny extra variance of about $5\%$.
* **Dimensionality Reduction:** Instead of keeping both height and weight, we can reduce them to one new feature ($PC1$) without losing much information.

## Applications and Limitations of PCA

### Applications of PCA

* **Finance:** Reduce correlated financial indicators to fewer risk factors.
* **Healthcare:** Compress genetic expression data into fewer factors.
* **Image Processing:** Face recognition (Eigenfaces).
* **Marketing:** Reduce survey question responses into main customer sentiment factors.

### Limitations of PCA

1. **Interpretability Issue**
* Principal components are linear combinations of original variables, making them harder to interpret meaningfully.
2. **Assumption of linearity**
* PCA only captures linear relationships, meaning non-linear structures in data are ignored.

3. **Sensitive to Scaling**
* If features are not standardized, PCA may give misleading results.


4. **Information Loss**
* Reducing dimensions means some variance (information) is lost.


5. **Impact of Outliers**
* Outliers can heavily influence principal components.

## Case Study (Finance Domain – Credit Risk)

* **Business problem:** Bank has $30+$ customer features (income, expenses, loans, credit history, etc.) $\rightarrow$ many correlated.
* **Goal:** Reduce to a smaller set of features for faster ML training.

### Approach:

* Apply PCA, retain $95\%$ variance with $8$ components instead of $30$.
* Train classification model (Default vs No Default) on reduced features.
* **Benefit:** Faster training, less noise, model generalizes better.
