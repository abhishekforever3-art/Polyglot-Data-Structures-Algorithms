## Hierarchical Clustering – Agglomerative with Dendrogram

### Introduction

Hierarchical clustering is a method of cluster analysis which seeks to build a hierarchy of clusters.

* **Dendrogram:** A tree that shows how clusters are merged step-by-step.
* **Cutting:** We cut the dendrogram at a certain height to form final clusters.

Agglomerative hierarchical clustering starts with each observation as its own cluster and iteratively merges the two closest clusters until all points are in a single cluster. The process is visually represented using a dendrogram.

## Step-by-Step Process of Dendrogram Creation

Agglomerative hierarchical clustering builds a bottom-up hierarchy of clusters through iterative merging. The step-by-step procedure unfolds as follows:

* **Initialization:** Each individual data point is treated as its own separate cluster.
* **Distance Calculation:** Compute the pairwise distances between all existing clusters.
* **Pair Selection:** Identify the two clusters that have the smallest distance between them.
* **Merging:** Merge the selected pair of clusters into a single new cluster.
* **Matrix Update:** Update the distance matrix to incorporate the newly formed cluster and remove the individual constituent clusters.
* **Iteration:** Repeat the process continuously until all data points are merged into one single overarching cluster.
* **Visualization:** The resulting dendrogram visually maps out the exact sequence and distance at which every merge occurs.
* **Cluster Extraction:** Making a horizontal cut across the dendrogram at a chosen height determines the final number of clusters.

## Linkage Criteria in Hierarchical Clustering

When performing agglomerative hierarchical clustering, the distance between two clusters determines which pairs get merged next. Different linkage methods define this inter-cluster distance in distinct ways:

* **Single Linkage (Nearest Neighbour):** The distance between two clusters is defined as the minimum distance between any single pair of points taken from each of the two clusters.
* **Complete Linkage (Farthest Neighbour):** The distance between two clusters is defined as the maximum distance between any single pair of points taken from each of the two clusters.
* **Average Linkage:** The distance between two clusters is defined as the average of all pairwise distances between the points belonging to the two respective clusters.

## Interpreting the Dendrogram

* **Leaves:** Represent the original data points.
* **Node Height:** Indicates the exact distance at which clusters were merged.
* **Horizontal Cut:** Cutting the dendrogram horizontally at a specific height defines the final number of clusters.
* **Merge Height:** A lower merge height reflects a stronger similarity between the merging clusters.

### Advantages of Hierarchical Clustering

* **No Pre-Specification Required:** Unlike K-Means, it does not require you to pre-define the number of clusters ($K$) beforehand.
* **Nested Hierarchy:** Produces a comprehensive, multi-level hierarchy of nested clusters represented visually by the dendrogram.

## Limitations and Practical Considerations of Hierarchical Clustering

### Limitations

* **Computationally Expensive:** Can be resource-intensive and slow when working with large datasets.
* **Sensitive to Noise and Outliers:** Anomalous data points can disproportionately impact distance calculations and skew early-stage merges.
* **Linkage Sensitivity:** Different linkage criteria (single, complete, or average) can produce entirely different cluster hierarchies.
* **Irreversible Merges:** Once two clusters are merged, that decision is permanent and cannot be undone or split later in the process.

### Practical Considerations

* **Feature Standardization:** Features should always be scaled or standardized before calculating Euclidean distance so that variables with larger numeric ranges don't dominate the results.
* **Compare Multiple Methods:** It is best practice to try multiple linkage methods to see how robust your clusters are.
* **Validation:** You can validate your hierarchical clustering results using metrics like the silhouette score or cophenetic correlation.


## Applications of Clustering

### 1. 🏛️ Banking & Finance

* **Customer Segmentation:** Group customers based on income, spending habits, or investment behavior.
* **Fraud Detection:** Identify unusual transactions by clustering normal transaction patterns.
* **Risk Management:** Segment loan applicants into risk categories.

### 2. 🛍️ Retail & E-commerce

* **Market Segmentation:** Cluster shoppers into categories (bargain hunters, premium buyers, etc.).
* **Product Recommendation:** Group products based on purchase patterns for cross-selling.
* **Store Clustering:** Identify stores with similar sales trends for regional marketing.

### 3. 🏥 Healthcare & Life Sciences

* **Patient Segmentation:** Group patients based on medical history, lifestyle, or treatment response.
* **Disease Classification:** Cluster genetic data or symptoms for disease subtypes.
* **Drug Discovery:** Group compounds with similar chemical structures.

### 4. 📱 Telecommunications

* **Churn Prediction:** Cluster customers by usage patterns to identify high risk churn groups.
* **Network Optimization:** Cluster similar usage zones for efficient resource allocation.
* **Service Personalization:** Design plans based on customer groups (heavy data users vs. casual callers).

### 5. 🌐 Marketing & Advertising

* **Targeted Campaigns:** Cluster audiences by demographics, interests, or behavior.
* **Brand Positioning:** Identify consumer perception groups from survey/feedback data.
* **Social Media Analysis:** Group influencers and audiences with similar engagement.
