Topic 1: Introduction to Forecasting & Time Series Analysis

**Definition**

Forecasting is the process of predicting future values based on historical data.

In time series forecasting, we have data collected at regular time intervals (daily, monthly, yearly, etc.), and our goal is to project future values by analyzing past patterns.

**Example:** Predicting next month's electricity consumption based on monthly usage for the past 3 years.

**Concept**

A **time series** is an ordered sequence of data points recorded over time.
When we forecast, we assume that:

* Past patterns can help us understand the future.
* Data may contain **trend**, **seasonality**, and **random fluctuations**.

*(Visual breakdown from left to right)*:


* **Positive Trend**: Steady upward slope over time
* **Seasonality**: Regular, repeating wave patterns over fixed intervals
* **Cyclic**: Longer-term wave patterns of ups and downs
* **Irregular**: Unpredictable spiky or random variations
* Some methods rely purely on past data (statistical), while others incorporate external factors (machine learning).

**Key idea:**
If you understand *what is driving* the changes in your series (trend, seasonal...)

events, special spikes), you can build a model that predicts future values with reasonable accuracy.

**Applications of Forecasting**

Forecasting is used across industries for:

1. **Business & Retail** — Sales forecasting, inventory planning, staffing schedules.
2. **Finance** — Stock prices, currency exchange rates, credit risk.
3. **Manufacturing** — Demand planning, machine maintenance schedules.
4. **Energy & Utilities** — Electricity load forecasting, water demand prediction.
5. **Healthcare** — Predicting patient inflow, disease spread.

6. **Transportation** — Traffic forecasting, airline passenger demand.

**Basic Math Example — Naïve Forecasting**

One of the simplest forecasting methods is the naïve method, where:

Forecast for next period = Value of last period

# Time Series Forecasting : - Naïve Forecasting (Simplest Forecasting method)

```python
import pandas as pd
import matplotlib.pyplot as plt

# Dataset
data = {
    "Month": ["Jan", "Feb", "Mar", "Apr"],
    "Sales": [200, 220, 240, None]  # April is unknown
}
df = pd.DataFrame(data)
df
```

```python

# Plot existing data
plt.plot(df["Month"], df["Sales"], marker='o', label="Actual Sales")

plt.title("Naïve Forecast Example")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)

# Naïve forecast
naive_forecast = df["Sales"].iloc[2]  # Last known value (March)

plt.plot("Apr", naive_forecast, marker='x', markersize=10, color='red', label="Forecast (Naïve)")

plt.legend()
plt.show()

print(f"Naïve Forecast for April: {naive_forecast}")
```

**Components of a Time Series**

**Definition**

A time series is made up of several underlying patterns that, when combined, produce the observed data. Breaking the series into its components helps us understand its behavior and select the right forecasting method.

**Concept**

There are four main components in a time series:

* **Level** — The baseline average value of the series, around which variations occur.
* **Trend** — The long-term upward or downward movement. Example: Steady sales growth over years.

* **Seasonality** — Regular, repeating patterns at fixed intervals. Example: Ice cream sales peaking every summer.
* **Noise (Irregular)** — Random fluctuations caused by unpredictable factors, such as sudden events or errors in data.

The combination of these components can be **additive** or **multiplicative**:

* **Additive model:**

$$Y_t = \text{Level} + \text{Trend} + \text{Seasonality} + \text{Noise}$$



Used when seasonal variations are roughly constant over time.
* **Multiplicative model:**

$$Y_t = \text{Level} \times \text{Trend} \times \text{Seasonality} \times \text{Noise}$$



Used when seasonal variations increase or decrease with the trend.

**Applications**

* **Business**: Identifying whether sales growth is due to long-term trend or seasonal spikes.
* **Economics**: Understanding if unemployment rate changes are due to cyclical trends or random shocks.
* **Energy**: Separating seasonal electricity demand from overall upward consumption trend.

**Numerical Example**

Imagine monthly sales for a shop:

* **Level**: The shop usually sells around 500 units/month.
* **Trend**: Sales increase by 10 units every month.
* **Seasonality**: Every December, sales go up by 50 units due to holiday shopping.

* **Noise**: Sometimes $+20$ or $-15$ units because of weather or local events.

For **December** in a year where baseline is $500$:

$$Y_t = 500 \text{ (Level)} + 10 \times 12 \text{ (Trend)} + 50 \text{ (Seasonality)} + 20 \text{ (Noise)}$$
