### 1. Introduction to Supervised Learning

#### What is Supervised Learning?

Supervised learning is a type of **machine learning** where a model is trained using **labeled data** — meaning the input data (features) is paired with the correct output (target/label).

* The goal is for the model to **learn a mapping function** from inputs to outputs so that it can predict the correct label for new, unseen data.
* The learning process involves adjusting model parameters to minimize prediction errors on the training data.

### Key Components of Supervised Learning:

1. **Input Features (X):** The independent variables used to make predictions (e.g., age, salary, pixel values).
2. **Output Labels (Y):** The dependent variable we want to predict (e.g., "spam" or "not spam," house price).
3. **Training Data:** A dataset with known (X, Y) pairs used to teach the model.
4. **Model:** The algorithm (e.g., linear regression, logistic regression, decision trees) that learns from the data.
5. **Loss Function:** Measures how wrong the model's predictions are (e.g., Mean Squared Error for regression, Log Loss for classification).
6. **Learning Algorithm:** Adjusts model parameters to minimize the loss (Error).

### Types of Supervised Learning Tasks

Supervised learning problems are broadly categorized into two types:

#### 1. Regression (Continuous Output)

* Predicts a **numerical value** (e.g., price, temperature, sales).
* **Examples:**
* Predicting house prices based on size, location, etc.
* Forecasting stock prices.


* **Common algorithms:**
* Linear Regression
* Polynomial Regression
* Decision Trees (for regression)

### 2. Classification (Discrete Output)

* Predicts a **category or class label** (e.g., "yes/no," "spam/not spam").
* **Examples:**
* Email spam detection (binary classification).
* Handwritten digit recognition (multiclass classification).

* Common algorithms:
* Logistic Regression
* Decision Trees
* Support Vector Machines (SVM)
* Random Forest



## Logistic regression is a classification algorithm (even though its name has "regression").

### What is Linear Regression?

Linear regression is a **supervised learning** algorithm used for predicting a **continuous numerical output** based on input features. It assumes a linear relationship between input variables ($X$) and the output ($Y$).

### How Does It Work?

1. **Training:** The model learns coefficients ($\beta$) that minimize the **Mean Squared Error (MSE)** between predicted and actual values.
2. **Prediction:** Once trained, it predicts $y$ for new inputs.

---

### Limitations of Linear Regression for Classification

* **Problem:** What if we want to predict a **binary outcome** (e.g., "Will a student pass or fail?")?
* Linear regression can output values outside $[0, 1]$, but probabilities must be between 0 and 1.

* **Example**: If $y = 1.5$ or $y = -0.3$, how do we interpret this as a probability?
* **Solution**: We need a function to **constrain predictions between 0 and 1** $\rightarrow$ Logistic Regression.

#### What is Classification?

Classification is a supervised learning task where the goal is to predict a **discrete class label** (category) rather than a continuous value.

### Types of Classification:

1. **Binary Classification (2 classes)**
* **Example:**
* Email: Spam (1) or Not Spam (0).
* Medical Test: Diseased (1) or Healthy (0).


* **Case Study: Credit Card Fraud Detection**
* **Input Features:** Transaction amount, location, time, etc.
* **Output:** Fraudulent (1) or Legitimate (0).
* **Model Used:** Logistic Regression (predicts probability of fraud).




2. **Multiclass Classification (>2 classes)**
* **Example:**
* Handwritten Digit Recognition (0-9).
* Animal Image Classification (Cat, Dog, Bird).

* Case Study: Iris Flower Classification
* **Input Features:** Sepal length, petal width, etc.
* **Output:** Setosa, Versicolor, or Virginica.
* **Model Used:** Multinomial Logistic Regression.



**Why Not Use Linear Regression for Classification?**

* **Issue 1:** Output is unbounded (can be $<0$ or $>1$), but probabilities must be in $[0, 1]$.

* **Issue 2:** Sensitive to outliers (a single extreme data point can skew predictions).
* **Solution:** Use logistic regression, which applies a **sigmoid function** to map predictions to $[0, 1]$.

---

### 4. Logistic Regression: The Concept From Linear to Logistic Regression

* **Linear Regression:** Predicts continuous $y$ using a straight line.
* **Logistic Regression:** Predicts probability $P(y=1\vert{}x)$ using an S-shaped curve (sigmoid).

### Sigmoid Function (Logistic Function)

$$P(y=1\vert{}x) = \frac{1}{1 + e^{-(\beta_0 + \beta_1 x_1 + \dots + \beta_n x_n)}}$$

#### Properties:

* Output always between 0 and 1.
* At $z = 0$, $P(y=1) = 0.5$.
* Approaches 1 as $z \to +\infty$, 0 as $z \to -\infty$.

---

### Example: Student Exam Pass/Fail Prediction

* **Problem:** Predict if a student passes (1) or fails (0) based on study hours.

* **Model Output:**
* If $P(y=1) \ge 0.5$, predict "Pass."
* Else, predict "Fail."



### Interpretation of Coefficients (Log-Odds)

* The equation inside the sigmoid is the log-odds:

$$\log\left(\frac{P(y=1)}{1 - P(y=1)}\right) = \beta_0 + \beta_1 x_1 + \dots + \beta_n x_n$$


* **Example:** If $\beta_1 = 1.5$ for study hours:
* A 1-hour increase $\to$ odds of passing multiply by $e^{1.5} \approx 4.48$.

### 5. Training Logistic Regression (Maximum Likelihood Estimation - MLE)

* Unlike linear regression (which uses Least Squares), logistic regression uses MLE to find coefficients.
* **Goal:** Choose $\beta$ values that maximize the likelihood of observing the actual data.

#### Example: Spam Detection

* **Model:** Predicts $P(\text{Spam}\vert{}\text{Email})$.
* **Training:** Adjusts weights to maximize the probability of correct classification.

---

### 6. Model Evaluation Metrics

Since accuracy alone can be misleading:




* **Metrics:**
* $\text{Accuracy} = \frac{\text{TP} + \text{TN}}{\text{TP} + \text{TN} + \text{FP} + \text{FN}}$
* **Precision:** $\frac{\text{TP}}{\text{TP} + \text{FP}}$ (How many predicted positives are correct?).

* **Recall:** $\frac{\text{TP}}{\text{TP} + \text{FN}}$ (How many actual positives were caught?).
* **F1-Score:** Harmonic mean of precision and recall.
* **ROC-AUC:** Measures model's ability to distinguish classes.

---

### Example: Student Pass/Fail Predictions (10 Cases)

Let's assume we have the following **Actual** and **Predicted** values:




### How to Calculate Metrics?

1. **Accuracy:**
* $(\text{TP} + \text{TN}) / \text{Total}$
* $= (4 + 2) / 10$
* $= 60\%$ (60% of predictions were correct)


2. **Precision (for "Pass"):**
* $\text{TP} / (\text{TP} + \text{FP})$
* $= 4 / (4 + 2)$
* $= 66.7\%$
* (When the model predicts "Pass," it's correct 66.7% of the time)


3. **Recall (Sensitivity):**
* $\text{TP} / (\text{TP} + \text{FN})$
* $= 4 / (4 + 2)$
* $= 66.7\%$
* (The model caught $66.7\%$ of actual "Pass" cases)

4. **F1-Score:**
* $2 \times \text{Precision} \times \text{Recall} / (\text{Precision} + \text{Recall})$
* $= 66.7\%$


5. **Assumptions & Limitations**

* **Linear Decision Boundary:** Assumes a linear relationship between features and log-odds.
* **No Multicollinearity:** Highly correlated features can distort coefficients.
* **Large Sample Size:** Works best with sufficient data.
* **Not for Complex Patterns:** Struggles with non-linear relationships (use SVM or Neural Networks instead).

```python
from sklearn.metrics import accuracy_score

# Get predictions for both train and test sets
y_pred_train = classifier.predict(x_train)
y_pred_test = classifier.predict(x_test)

# Calculate accuracies
train_accuracy = accuracy_score(y_train, y_pred_train)
test_accuracy = accuracy_score(y_test, y_pred_test)

print(f"Train Accuracy: {train_accuracy:.3f}")
print(f"Test Accuracy: {test_accuracy:.3f}")

# Determine generalization status
if abs(train_accuracy - test_accuracy) < 0.05:
    print("✅ Good generalization (similar accuracies)")
elif train_accuracy > test_accuracy + 0.1:
    print("⚠️ Overfitting detected (train >> test). Solutions:")
    print("- Add regularization (e.g., LogisticRegression(penalty='l2', C=0.1))")
    print("- Reduce features (select top features using RFE or feature_importances_)")
    print("- Increase training data or use cross-validation")
else:
    print("❓ Check model or data (unexpected gap)")
```

### Key Enhancements
### 1. Regularization (L2 Penalty)

### What?

A technique to prevent overfitting by penalizing overly complex models. L2 regularization (aka "Ridge") shrinks coefficient values toward zero but never fully removes features.

### Why?

When your model performs well on training data but poorly on test data (overfitting), L2 regularization helps by:

Reducing the impact of less important features.

Keeping all features but with smaller weights.

```python
# Improved model with regularization
classifier = LogisticRegression(penalty='l2', C=0.1, solver='liblinear', random_state=42)
classifier.fit(x_train, y_train)
```

```python
# Why?

# penalty='l2' reduces overfitting by penalizing large coefficients.

# C=0.1 increases regularization strength (lower C = stronger penalty).
```

### 2 Feature Selection (RFE)

### What?

Recursive Feature Elimination (RFE) selects the top n most important features by recursively removing the least significant ones.

### Why?
Improves generalization by eliminating noisy/irrelevant features.

Reduces training time and model complexity.

```python
from sklearn.feature_selection import RFE

# Select top 3 features
selector = RFE(classifier, n_features_to_select=3, step=1)
selector.fit(x_train, y_train)
print("Selected features:", x_train.columns[selector.support_])
```

```python
#Removes irrelevant features that may cause overfitting.
```

### 3. Cross-Validation

### What?

A method to evaluate model performance by splitting data into multiple train/test folds (e.g., 5-fold).

### Why?

Provides a robust estimate of generalization error.

Reduces reliance on a single train-test split.

```python
from sklearn.model_selection import cross_val_score

# 5-fold cross-validation
cv_scores = cross_val_score(classifier, x_train, y_train, cv=5, scoring='accuracy')
print(f"Cross-validated Accuracy: {cv_scores.mean():.3f} (±{cv_scores.std():.3f})")
```
