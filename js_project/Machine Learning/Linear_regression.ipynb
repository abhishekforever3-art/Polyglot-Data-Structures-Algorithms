### LINEAR REGRESSION - CONCEPT

Let's understand what these 2 terms mean independently.

* **Linear:** this means progressing from one stage to another in a single series of steps (sequential) extending along a straight or nearly straight line.
* **Regression:** The statistical terminology for regression is to determine the relationship between one dependent variable and a series of independent variables.

Thus, *linear regression* means a linear approach for modeling the relationship between dependent and independent quantitative variables. It will plot a straight line known as a best-fit line along points to predict the target value.

### PURPOSE OF LINEAR REGRESSION

* The purpose of linear regression is to predict or forecast value bases on input.
* So in linear regression, we have input and output. Input is termed as independent variable and output is dependent variable.
* Example in house price prediction, we have two features (variables) number of rooms and price, one is dependent variable and another is independent variable, so based on problem definition.

### TYPES OF LINEAR REGRESSION

The linear regression line has an equation: $y = \beta_0 + \beta_1x$ ($Y = mX + c$) where $Y$ is the dependent variable, $X$ is the independent variable, $\beta_1$ ($m$) is the slope, and $\beta_0$ ($c$) is the intercept (the value of $y$ when $x = 0$).

There are 2 types of Linear Regression models:

* **Simple Linear Regression:** It is a linear regression model with one independent variable and one dependent variable: $y = \beta_0 + \beta_1x_1$
* **Multiple Linear Regression:** It is a linear regression model with more than one independent variable and one dependent variable: $y = \beta_0 + \beta_1x_1 + \beta_2x_2 + \dots$

### HOW TO CALCULATE COEFFICIENT B0 AND B1

**Ordinary Least Squares (OLS):**

* This is the most common method. It minimizes the sum of the squares of the differences between the observed and predicted values.
* The formulas for $b_0$ and $b_1$ are derived by minimizing the sum of the squared differences between the observed and predicted values.

---

### Formulas

$$b_1 = \frac{N(\sum xy) - (\sum x)(\sum y)}{N(\sum x^2) - (\sum x)^2}$$

$$b_0 = \frac{\sum y - b_1(\sum x)}{N}$$

---

### Where:

* **$N$**: Number of data points (in this case, 6)
* **$\sum x$**: Sum of the number of rooms
* **$\sum y$**: Sum of the prices
* **$\sum xy$**: Sum of the product of the number of rooms and the corresponding prices
* **$\sum x^2$**: Sum of the squares of the number of rooms

### Linear Regression Assumptions

1. **Linearity**
* *Description:* Linear relationship between $Y$ and each $X$.
* *Correct:* A straight trend line through data points (marked with a green check).
* *Incorrect:* A curved or non-linear relationship (marked with a red cross).


2. **Homoscedasticity**
* *Description:* Residuals have constant variance across all predicted values.
* *Correct:* Even spread of residuals along the regression line (marked with a green check).
* *Incorrect:* A funnel-shaped or widening spread of residuals (marked with a red cross).


3. **Multivariate Normality**
* *Description:* Residuals should be normally distributed around the regression line.
* *Correct:* A proper normal distribution curve under the residuals (marked with a green check).
* *Incorrect:* A non-normal or skewed distribution (marked with a red cross).


4. **Independence**
* *Description:* Independence of observations (includes "no autocorrelation").
* *Correct:* Independent data point distribution (marked with a green check).
* *Incorrect:* Wavy or autocorrelated residual patterns (marked with a red cross).


5. **Lack of Multicollinearity**
* *Description:* Predictors are not correlated with each other.
* *Correct:* $X_1 \neq X_2$ (marked with a green check).
* *Incorrect:* $X_1 \sim X_2$ (marked with a red cross).


6. **The Outlier Check**
* *Description:* Extreme values that may disproportionately influence the regression results.
* *Correct:* Well-behaved data points following the regression trend line (marked with a green check).

### Import Data Set

```python
import pandas as pd

df_news = pd.read_csv('NewspaperData.csv')
```

### Fitting a liner regression model to predict Sunday newspaper circulation based on daily newspaper circulation

```python
import statsmodels.formula.api as smf
model = smf.ols("sunday~daily", data=df_news).fit()
# ols = ordinary least squares it is a method for estimating the unknown parameters in a linear regression model. OLS chooses the parameters of a linear function of a set of explanatory variables by the principle of least squares: minimizing the sum of the squares of the differences between the observed dependent variable in the given dataset and those predicted by the linear function.
# fit() method is used to fit the model to the data and estimate the parameters of the linear regression model.
#fit is a method that is used to train the model on the given data. It takes the input data and the target variable and finds the best-fitting line that minimizes the difference between the predicted values and the actual values.
```

```python
model.params
```

```python
# t and p values are used to determine the statistical significance of the coefficients in a regression model. The t-value measures how many standard deviations the coefficient is away from zero, while the p-value indicates the probability of observing a t-value as extreme as the one calculated, assuming the null hypothesis is true. A low p-value (typically < 0.05) suggests that the coefficient is statistically significant and not likely due to random chance.
print(f'model t values:\n{model.tvalues}\n')
print("----------------------------------")
print(f'model p values:\n{model.pvalues}\n')
```


### R-squared

* $R^2$ tells us how well the model fits the data. Closer to $1 =$ better fit.
* Example: If $R^2 = 0.91$, the model explains 91% of the variation in $y$.
* But $R^2$ has a quirk—it always increases when we add more features, even useless ones!

---

### Adjusted R-squared deals with this issue.

* Adjusted $R^2$ fixes this quirk by penalizing extra features that don't help.
* Rule of thumb: If a feature doesn't improve the model enough, Adjusted $R^2$ decreases.
* Watch how Adjusted $R^2$ changes when we add/remove features later!


### Predicting Sales Based on Store Footfalls using Simple Linear Regression

We want to predict store sales based on footfall using a simple linear regression model.

Just like more people in a store usually means more sales, we'll use this pattern to predict future sales.

---

### Linear Regression Overview

Linear Regression is a supervised learning algorithm used to model the relationship between one or more independent variables (features) and a continuous dependent variable (target).

**Types of Linear Regression:**

#### Simple (or Single) Linear Regression:

Uses one independent variable to predict a continuous target variable by fitting a straight line.

## Multiple Linear regression:
uses tow or more variables to predict the target by fitting a multi-dimentional hyperplane.

```python
import pandas as pd 
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
data = pd.read_csv('General_Stores.csv')
print("Data Preview:")
data.head()
```

Define independent (footfalls) and dependent (dailySales) variables

```python
x = data[['Footfalls']]
y = data['DailySales']
```

* Double brackets (`[['Footfalls']]`) are used to select a DataFrame with two dimensions (rows and columns).
* Scikit-learn expects the input features ($X$) to be in a 2D array-like structure, even if there is only one feature.

---

### Y (Target / Output)

* Single brackets (`['DailySales']`) are used to select a Series, which is a 1D structure (just one column).
* The target variable ($Y$) should be a 1D array or Series.

## Create and train linear regression model

### Simple Linear Regression Equation

$$y = \beta_0 + \beta_1x$$

### Where:

* $y$ = Predicted value (dependent variable, e.g., sales)
* $x$ = Input value (independent variable, e.g., footfall)
* $\beta_1$ = Slope (coefficient showing how much $y$ changes with $x$)
* $\beta_0$ = Intercept (value of $y$ when $x = 0$)
